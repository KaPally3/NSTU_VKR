```python
%matplotlib widget
# Моделирование мультипризматических рентгеновских линз: влияние геометрии линзы на её оптические свойства
import sys
import numpy as np
import argparse
import matplotlib.pyplot as plt

res = 4000
xmin, xmax = -0.2, 0.2

PI = np.pi
EN = 30
LAMBDA = 1e-6 * 1239.8 / (EN * 1e3)
K = 2 * PI / LAMBDA
MU = 3.0235443337462318
BETA = MU * LAMBDA / (4 * PI)
DELTA = 5.43e-4 / (EN**2)
FOCAL = 8000
THETA = 130

source_lens_distance = 2 * FOCAL
#lens_screen_distance = 2 * FOCAL

xs = np.linspace(xmin, xmax, res)
dx = (xmax - xmin) / (res - 1)
fx = np.fft.fftfreq(res, d=dx)
fx = np.fft.fftshift(fx)
lens_thickness = xs**2 / (2 * FOCAL * DELTA)
formula_one = (
    EN
    * np.exp(1j * K * np.sqrt(xs**2 + source_lens_distance**2))
    / np.sqrt(xs**2 + source_lens_distance**2)
    * np.exp(-1j * K * (DELTA - 1j * BETA) * lens_thickness)
)
fig, ax1 = plt.subplots()
ax2 = ax1.twinx()
ax1.plot(xs, np.abs(formula_one))
ax2.plot(xs, np.angle(formula_one))
plt.show()

for lens_screen_distance in np.arange (1.5 * FOCAL, 2.6 * FOCAL, 0.1 * FOCAL):
    x_screen = fx * LAMBDA * lens_screen_distance
    formula_two = np.exp(PI * 1j / (LAMBDA * lens_screen_distance) * (xs**2))

    formula_three = (
        -1j
        / LAMBDA
        * np.exp(
            2
            * PI
            * 1j
            / LAMBDA
            * (lens_screen_distance + (x_screen**2) / 2 / lens_screen_distance)
        )
    )

    out_source = np.fft.fft(formula_one * formula_two)
    out_source = formula_three * np.fft.fftshift(out_source)

    fig, ax1 = plt.subplots()
    ax2 = ax1.twinx()
    ax1.plot(x_screen, np.abs(out_source))
    ax2.plot(x_screen, np.angle(out_source))
    plt.xlim(-0.04,0.04)
    plt.text(0,4,str(lens_screen_distance))
plt.show()
```

```python
from scipy.optimize import curve_fit

def f(x, a, c, s):
    return a * np.exp(-.5 * ((x - c) / s)**2)

p0 = [2e7, 0, .1]

popt, pcov = curve_fit(f=f, xdata=x_screen, ydata=np.abs(out_source), p0=p0)
print(popt)

fig, ax1 = plt.subplots()
ax1.plot(x_screen, np.abs(out_source), "+")
ax1.plot(x_screen, f(x_screen, *p0))
ax1.plot(x_screen, f(x_screen, *popt))
plt.xlim(-.7, .7)
```

```python
from scipy.interpolate import make_interp_spline

distances = []
amplitude_values = []

def gaussian(x, a, c, s):
    return a * np.exp(-0.5 * ((x - c) / s) ** 2)

for lens_screen_distance in np.arange(1.5 * FOCAL, 2.6 * FOCAL, 0.1 * FOCAL):
    x_screen = fx * LAMBDA * lens_screen_distance
    formula_two = np.exp(PI * 1j / (LAMBDA * lens_screen_distance) * (xs**2))
    formula_three = (
        -1j
        / LAMBDA
        * np.exp(
            2 * PI * 1j / LAMBDA * (lens_screen_distance + (x_screen**2) / (2 * lens_screen_distance))
        )
    )

    out_source = np.fft.fft(formula_one * formula_two)
    out_source = formula_three * np.fft.fftshift(out_source)
    amplitude = np.abs(out_source)

    p0 = [np.max(amplitude), 0, 0.01] 
    popt, pcov = curve_fit(gaussian, x_screen, amplitude, p0=p0)
    half_width_amp = 2.355 * abs(popt[2])  

    
    distances.append(lens_screen_distance)
    amplitude_values.append(half_width_amp)

print("Distance | half_width_amp")
for d, f in zip(distances, amplitude_values):
    print(f"{d:} | {f:.6f}")

plt.figure()
plt.plot(distances, amplitude_values)
plt.xlabel("Lens Screen Distance")
plt.grid()
plt.show()

#interpolation
x_smooth = np.linspace(min(distances), max(distances), 300)
y_smooth = make_interp_spline(distances, amplitude_values, k=3)(x_smooth)

plt.figure()
plt.plot(x_smooth, y_smooth)
plt.xlabel("Lens Screen Distance")
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def generate_crocodile(L, N, y_t, y_g, extra_t, y_lines):
    teeth = []
    step = L / N
    tilt_radians = np.arcsin(y_g / L)
    cos_theta = np.cos(tilt_radians)
    sin_theta = np.sin(tilt_radians)
    
    all_distances = np.zeros((len(y_lines), N))
    
    for i in range(N):
        x1 = i * step
        x2 = x1 + step / 2
        x3 = x1 + step
        
        y1 = extra_t
        y2 = extra_t + y_t
        
        tooth_x = np.array([x1, x2, x3, x3, x1, x1])
        tooth_y = np.array([y1, y2, y1, y1 - extra_t, y1 - extra_t, y1])
        
        x_centered = tooth_x - (L / 2)
        y_centered = tooth_y
        
        # Rotation
        tooth_x_rotated = cos_theta * x_centered - sin_theta * y_centered + (L / 2)
        tooth_y_rotated = sin_theta * x_centered + cos_theta * y_centered
        
        teeth.append((tooth_x_rotated, tooth_y_rotated))
    
    # Корректировка высоты зубов относительно последнего зуба
    min_shift = teeth[-1][1][1]
    for i in range(len(teeth)):
        teeth[i] = (teeth[i][0], teeth[i][1] - min_shift)
    
    for idx, y_line in enumerate(y_lines):
        for i, (tooth_x_rotated, tooth_y_rotated) in enumerate(teeth):
            intersections = []
            for j in range(len(tooth_x_rotated) - 1):
                x_start, x_end = tooth_x_rotated[j], tooth_x_rotated[j + 1]
                y_start, y_end = tooth_y_rotated[j], tooth_y_rotated[j + 1]
                
                if (y_start - y_line) * (y_end - y_line) <= 0:
                    x_intersect = x_start + (x_end - x_start) * ((y_line - y_start) / (y_end - y_start))
                    intersections.append(x_intersect)
            
            if len(intersections) == 2:
                all_distances[idx, i] = abs(intersections[1] - intersections[0])
            else:
                all_distances[idx, i] = 0
    
    return teeth, all_distances

L = 80 
N = 10 
y_t = 5
y_g = 1 
extra_t = 1 
y_lines = np.linspace(0, -y_t, 5) 

teeth, distances = generate_crocodile(L, N, y_t, y_g, extra_t, y_lines)

plt.figure(figsize=(8, 4))
for tooth_x, tooth_y in teeth:
    plt.plot(tooth_x, tooth_y)
for y_line in y_lines:
    plt.axhline(y_line, color='r', linestyle='--', alpha=0.5)
plt.xlabel("X [mm]")
plt.ylabel("Y [mm]")
plt.grid()
plt.show()

plt.figure(figsize=(8, 4))
colors = plt.cm.viridis(np.linspace(0, 1, N))
for i in range(N):
    plt.plot(y_lines, distances[:, i], color=colors[i], label=f'{i+1}')
plt.legend()
plt.grid()
plt.show()


print(distances)
```
